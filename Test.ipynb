```python
import sys
import os
import pytest

sys.dont_write_bytecode = True
os.environ["PYTHONDONTWRITEBYTECODE"] = "1"

# ================= Configure Paths =================
src_path = "/Workspace/9900-f18a-cake/working_branch/src"
tests_path = "/Workspace/9900-f18a-cake/working_branch/src/mch/tests"

# ================= Set Environment =================
if src_path not in sys.path:
    sys.path.append(src_path)
    print(f"Path appended: {src_path}")

# ================= Run Tests =================

file1 = os.path.join(tests_path, "test_scheduler.py")
file2 = os.path.join(tests_path, "test_job_submission.py")
print(f"Running ONLY new tests...")
exit_code = pytest.main([file1, file2, "-v", "-p", "no:cacheprovider"])

# ================= Results Summary =================
if exit_code == 0:
    print("\n \033[92mSUCCESS: All tests passed! Great job!\033[0m")
elif exit_code == 1:
    print("\n \033[91mFAILURE: Some tests failed. Check the logs above.\033[0m")
else:
    print(f"\n Pytest exited with code: {exit_code}")
```
